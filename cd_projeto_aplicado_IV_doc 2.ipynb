## 1. Proposta do Projeto

### 1.1. Contexto

As séries temporais são amplamente empregadas na climatologia, na gestão pública e, mais recentemente, na avaliação
de risco de desastres, permitindo caracterizar o comportamento de um fenômeno ao longo do tempo e apoiar decisões
futuras a partir dele (MORETTIN; TOLOI, 2018). Este projeto insere-se na área de Ciência de Dados aplicada à Gestão
de Risco de Desastres e ao Planejamento Urbano, dedicando-se à análise da série temporal de focos de queimadas no
território brasileiro — dado histórico monitorado pelo Instituto Nacional de Pesquisas Espaciais através do Programa
Queimadas (INPE, 2026).

O projeto está alinhado ao Objetivo de Desenvolvimento Sustentável 11 (ODS 11) — Cidades e Comunidades
Sustentáveis —, cuja meta 11.5 propõe reduzir significativamente o número de mortes e perdas econômicas causadas
por desastres, e cuja meta 11.b trata do desenvolvimento do gerenciamento holístico do risco de desastres em todos
os níveis de governo (IPEA, [s.d.]). Nesse sentido, o problema selecionado consiste na dificuldade de antecipar a
distribuição temporal dos focos de calor ao longo dos meses e anos, de forma a apoiar a atuação de órgãos de defesa
civil e brigadas municipais de incêndio.

### 1.2. Motivação

Estudos anteriores já demonstraram a viabilidade de modelar estatisticamente o número de focos de calor. Espinosa,
Prado e Ghellere (2010) aplicaram um modelo SARIMA para prever focos de calor no Estado de Mato Grosso com um ano
de antecedência, encontrando resultados estatisticamente adequados para os meses mais críticos do ano. Viganó et
al. (2017), por sua vez, utilizaram Redes Neurais Artificiais para prever o número de focos e a área queimada no
Pantanal Sul-Mato-Grossense, obtendo bons níveis de ajuste ao utilizar a própria série histórica como preditora.

Esses estudos, no entanto, concentram-se em recortes regionais específicos — um estado ou um bioma — e adotam uma
lente predominantemente ambiental, sem conectar explicitamente a previsão ao planejamento de resposta a desastres
em nível municipal, tal como preconizado pela meta 11.b do ODS 11. Essa é a lacuna que este projeto busca abordar:
caracterizar a distribuição temporal dos focos em nível nacional como insumo direto para o planejamento logístico
de órgãos públicos responsáveis pela resposta a desastres, e não apenas como um retrato do fenômeno ambiental em
si.

### 1.3. Objetivo Geral

Desenvolver um produto analítico de previsão e caracterização da série temporal de focos de queimadas no Brasil,
como apoio à gestão de risco de desastres em nível municipal (ODS 11).

### 1.4. Objetivos Específicos

- Caracterizar a série temporal do número de focos de calor, identificando tendência, sazonalidade e ruído;
- Definir e comparar técnicas de modelagem estatística e/ou de aprendizado de máquina adequadas à série;
- Construir um modelo preditivo do volume de focos de calor futuro, fornecendo métricas de suporte à decisão para
  órgãos de defesa civil e brigadas de incêndio.

### 1.5. Justificativa

A antecipação de cenários futuros de incidência de queimadas contribui diretamente para a meta 11.5 do ODS 11 ao
subsidiar o planejamento logístico de brigadas de incêndio e a alocação de recursos públicos em períodos de maior
risco (IPEA, [s.d.]). Ao disponibilizar publicamente o código e os resultados deste projeto, busca-se também
favorecer a replicação da metodologia por gestores municipais com recursos analíticos limitados, atendendo ao
caráter extensionista do componente curricular.

## 2. Descrição da Base de Dados

A base de dados selecionada provém do banco de dados abertos do Instituto Nacional de Pesquisas Espaciais (INPE),
por meio do Programa Queimadas (INPE, 2026). Trata-se de um conjunto de dados reais e públicos, amplamente
utilizado como referência analítica em estudos correlatos (ESPINOSA; PRADO; GHELLERE, 2010; VIGANÓ et al., 2017).

**Informações disponíveis e estrutura:** a amostragem original consiste em dados diários com granularidade a nível
de foco de calor individual. A análise do conjunto de dados bruto demonstra a presença de diversas variáveis
estruturadas da seguinte forma:

| **Atributo** | **Descrição / Tipo** |
| --- | --- |
| data_hora_gmt | Data e hora da detecção do foco, fundamental para a ordenação dos registros no tempo. |
| lat e lon | Coordenadas geográficas precisas do evento (latitude e longitude). |
| satelite | Identificação do sensor de captura de imagem (ex: satélite GOES-19). |
| municipio, estado, bioma | Dados categóricos e de agrupamento espacial, essenciais para o recorte municipal proposto. |
| precipitacao, risco_fogo, numero_dias_sem_chuva, frp | Fatores meteorológicos associados e poder radiativo do fogo no instante da detecção. |

**Forma e período de coleta:** os dados são coletados de maneira contínua e regular por meio de sensoriamento
remoto via satélite (INPE, 2026). Para fins da construção da série temporal univariável neste projeto, esses
registros pontuais e irregulares serão agrupados temporalmente (soma diária, semanal ou mensal do volume de
focos por município), originando uma sequência ordenada de observações igualmente espaçadas ao longo do tempo —
mesma abordagem de agregação empregada por Espinosa, Prado e Ghellere (2010).

**Fonte dos dados:** Portal INPE — Estatísticas e Dados Abertos do Programa Queimadas (INPE, 2026).
